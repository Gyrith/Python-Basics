```python
from datetime import datetime, timedelta

today = datetime.now()

print(today + timedelta(days=7))
```

```python
print(today - timedelta(days=30))
```

```python
print(today + timedelta(hours=5))
```

# Lessons
 - The `datetime` module works with dates and times.
 - `date.today()` returns the current date.
 - `datetime.now()` returns the current date and time.
 - `strftime()` formats dates and times.
 - `timedelta` is used to add or subtract time.
 - The `calendar` module helps display calendars and check leap years

## Common Mistakes

- Forgetting to import `datetime` or `calendar`.
- Confusing `date` with `datetime`.
- Using the wrong `strftime()` format codes.
- Forgetting that `timedelta` requires keyword arguments like `days=` or `hours=`.
