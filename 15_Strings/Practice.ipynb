```python
#String formatting (f-strings)
name = "Grace"
age = 23

print(f"My name is {name} and I am {age} years old.")
```

```python
item = "Laptop"
price = 50000

print(f"The {item} costs KSh {price}.")
```

```python
a = 10
b = 20

print(f"{a} + {b} = {a + b}")
```

```python
#count vowels
text = input("Enter a word: ")

count = 0

for letter in text.lower():
    if letter in "aeiou":
        count += 1

print("Number of vowels:", count)
```

```python
word = input("Enter a word: ")

if word.lower() == word[::-1].lower():
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
password = input("Enter a password: ")

if len(password) >= 8:
    print("Strong password")
else:
    print("Password must be at least 8 characters long.")
```

# Lessons
 - Strings are sequences of characters.
 - I can access characters using indexes.
 - Slicing extracts parts of a string.
 - String methods help modify and analyze text.
 - I can split and join strings.
 - f-strings make formatting easy and readable.

# Common mistakes
 - Forgetting that string indexing starts at 0.
 - Using an index that is out of range.
 - Thinking that methods like `upper()` change the original string automatically.
 - Forgetting to use `f` before an f-string.
