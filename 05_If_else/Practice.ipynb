```python
#If...else
age = 16
if age >= 18:
    print('You can vote.')
else:
    print('You cannot vote.')
```

```python
number = 9
if number % 2 == 0:
    print('Even')
else:
    print('Odd')
```

```python
password = 'Python'
if password == 'Python':
    print('Correct password')
else:
    print('Wrong password')
```

```python
#if...elif...else
marks = 85
if marks >= 80:
    print('Grade A')
elif marks >= 70:
    print('Grade B')
else:
    print('Grade C')
```

```python
month = 12
if month == 12:
    print('December')
elif month == 1:
    print('January')
else:
    print('Another month')
```

```python
num = 0
if num > 0:
    print('Positive')
elif num < 0:
    print('Negative')
else:
    print('Zero')
```

```python
#Nested if
age = 22
has_id = True
if age >= 18:
    if has_id:
        print('Entry allowed.')
```

```python
username = 'admin'
password = '3456'
if username == 'admin':
    if password == '3456':
        print('Login successful.')
 
```

```python
balance = 5000
withdraw = 3000
if withdraw <= balance:
    if withdraw > 0:
        print('Transaction approved.')
```

```python
#Grade calculator
marks = float(input('Enter your marks:'))
if marks >= 80:
    print('Grade: A')
elif marks >= 70:
    print('Grade: B')
elif marks >= 60:
    print('Grade: C')
elif marks >= 50:
    print('Grade: D')
else:
    print('Grade: F')
```

```python
#ATM withdrawal
balance = 5000
print('Current Balance:',balance)
amount = float(input('Enter amount to withdraw:'))

if amount <= 0:
    print('Please enter a valid amount.')
elif amount <= balance:
    balance -= amount
    print('Withdrawal successful!')
    print('Remaining Balance:',balance)
else:
    print('Insufficient funds.')
```
