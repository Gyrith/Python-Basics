# Tuples
## Objectives
 -Creating tuples.
 -Accessing tuple elements
 -Tuple methods
 -Tuple packing and unpacking
 -Looping through tuples.
 -Nested tuples

```python
#Creating tuples
fruits = ('Apple','Banana','Orange')
print(fruits)
```

```python
numbers = (10,20,30,40,50)
numbers
```

```python
mixed = ('Grace',23,True,5.7)
mixed
```

Tuples use parantheses() and cannot be modified after creation.(they are immutable)

```python
#Accessing tuple elements
fruits = ('Apple','Banana','Orange')
fruits[0]
```

```python
fruits[1]
```

```python
#Tuple methods
#count method
numbers = (1,2,3,4,5,6,2)
print(numbers.count(2))
```

```python
#index method
fruits = ('Apple','Banana','Orange')
fruits.index('Banana')
```

```python
colors = ('Red','Blue','Green','Blue')
colors.count('Blue')
```

```python
#Tuple packing and unpacking
person = ('Grace',23,'Kenya',1.45)
person
```

```python
#unpacking
name,age,country,height = person
print(name)
print(age)
print(country)
```

```python
student = ('John',20,'Data Science')
name,age,course = student
print(course)
```

```python
#Looping through tuples
fruits = ('Apple','Banana','Orange')
for fruit in fruits:
    print(fruit)
```

```python
numbers = (10,20,30,40)
for number in numbers:
    print(number)
```

```python
for letter in ('P','Y','T','H','O','N'):
    print(letter)
```

```python
#Nested tuples
students = (
    ('Grace',23),
    ('John',21),
    ('Mary',22)
)
students
```

```python
print(students[0])
```

```python
students[1][0]
```

```python
student = ('Grace',23,'Data Science','Kenya')
name,age,course,country = student
print("Name:", name)
print("Age:", age)
print("Course:", course)
print("Country:", country)
```

```python
sales = (1200, 1500, 1800, 2000, 1700)

print("Highest Sale:", max(sales))
print("Lowest Sale:", min(sales))
print("Total Sales:", sum(sales))
print("Average Sale:", sum(sales) / len(sales))
```

```python
products = (
    ("Laptop", 5),
    ("Mouse", 15),
    ("Keyboard", 10)
)

for product, quantity in products:
    print(f"{product}: {quantity} in stock")
```

# Lessons
 -Tuples are ordered collections
 -Tuples are immutable (cannot be changed).
 -Tuples support indexing and slicing.
 -Tuples can be packed and unpacked.
 -Tuples can contain other tuples.

# Common mistakes
 -Trying to change a tuple element.
 -Confusing tuples '()' with lists'[]'.
 -Unpacking into wrong number of variables.
 -Forgetting that tuple
