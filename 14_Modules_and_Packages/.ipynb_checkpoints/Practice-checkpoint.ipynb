A package is a folder that contains mltiple python modules

```python
import random
import string

characters = string.ascii_letters + string.digits

password = ""

for i in range(8):
    password += random.choice(characters)

print(password)
```

```python
import random

for i in range(5):
    print(random.randint(1, 6))
```

```python
from datetime import date

birth_year = int(input("Enter your birth year: "))

current_year = date.today().year

age = current_year - birth_year

print("Your age is", age)
```

# Lessons
- A module is a Python file containing reusable code.
- Python provides many built-in modules.
- I can import an entire module or specific functions.
- Aliases make module names shorter.
- I can create my own modules.
- Packages organize related modules into folders.

#Common mistakes
 - Forgetting to `import` a module before using it.
 - Misspelling a module or function name.
 - Naming your own file the same as a built-in module (for example, `math.py`), which can cause import problems.
 - Forgetting that your custom module (`calculator.py`) must be in the same folder as your notebook, or in a location Python can find.
