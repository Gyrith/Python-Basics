```python
#for loop
for i in range(5):
    print(i)
```

```python
fruits = ['Apple','Banana','Orange']
for fruit in fruits:
    print(fruit)
```

```python
name = 'Python'
for letter in name:
    print(letter)
```

```python
#While loop
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
number = 10
while number >= 1:
    print(number)
    number -= 1
```

```python
password = ""
while password != 'python123':
    password = input('Enter password:')
print('Access Granted!')
```

```python
#Range()
for i in range(10):
    print(i)
```

```python
for i in range(5,11):
    print(i)
```

```python
for i in range(0,21,2):
    print(i)
```

```python
#Break
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
num = [2,4,6,8,10]
for number in num:
    if number == 8:
        break
    print(number)
```

```python
while True:
    word = input("Type ' quit' to stop:")
    if word == 'quit':
        break
    print(word)
```

```python
#Continue
for i in range(6):
    if i == 3:
        continue
    print(i)
```

```python
for letter in 'Python':
    if letter == 'h':
        continue
    print(letter)
```

```python
for number in range(1,11):
    if number % 2 == 0:
        continue
    print(number)
```

```python
#Pass
for i in range(5):
    pass
print('Loop finished.')
```

```python
if True:
    pass
print('Program continues.')
```

```python
def future_function():
    pass
```

```python
#Nested loops
for i in range(3):
    for j in range(3):
        print(i,j)
```

```python
for row in range(3):
    for col in range(4):
        print('*',end='')
    print()
```

```python
for i in range(1,4):
    for j in range(1,4):
        print(i*j,end='')
    print()
```

```python
total = 0
for i in range(1,101):
    total += i
print('Sum:',total)
```

```python
secret = 7
while True:
    guess = int(input('Guess the number:'))
    if guess == secret:
        print('Congratulations!You guessed correctly.')
        break
    else:
        print('Try again.')
```

```python
rows = 5
for i in range(1,rows + 1):
    print('*' * i)
```

# Lessons from loops
-'for' loops represent a block of code over a sequence.
-'while' loops repeat as long as a condition is true.
-'range()' generates a sequence of numbers.
-'break' exits a loop immediately.
-'continue' skips the current iteration.
-'pass' acts as a placeholder.
-Nested loops are loops inside other loops.

## Common Mistakes
-Forgetting to update the counter in a 'while' loop,causing an infinite loop.
-Using 'break' when you meant 'continue'.
-Incorrect identation inside loops.
-Using the wrong arguments with 'range()'.
