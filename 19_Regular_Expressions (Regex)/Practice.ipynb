```python
#pattern matching
import re

text = "My number is 12345"

print(re.findall(r"\d+", text))
```

```python
#validating data
import re

email = "student@gmail.com"

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```

```python
import re

phone = "0712345678"

pattern = r'^\d{10}$'

if re.match(pattern, phone):
    print("Valid Phone Number")
else:
    print("Invalid Phone Number")
```

```python
import re

password = "Python123"

pattern = r'^(?=.*[A-Za-z])(?=.*\d).{8,}$'

if re.match(pattern, password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
import re

text = "Contact us at support@gmail.com"

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

print(emails)
```

# Lessons
 -Regex is used to search and manipulate text.
 -'re.search()' finds the first match.
 -'re.findall()' returns all matches.
 -'re.sub()' replaces matching text.
 -'re.split()' splits text using a pattern.
 -Regex can validate emails,phone numbers and passwords.

# Common mistakes
 - Forgetting to `import re`.
 - Calling `.group()` when `re.search()` returns `None`.
 - Using the wrong regex pattern.
 - Forgetting to use raw strings (`r"pattern"`) for regex patterns
