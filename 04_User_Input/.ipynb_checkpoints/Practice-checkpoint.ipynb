# User Input
## Objectives
I will learn:
 -How to use input()
 -Store user input in variables
 -Convert input to int and float
 -Build simple interactive progress

```python
#Basic input
name = input('Enter your name:')
print('Hello,',name)
```

```python
#multiple inputs
name = input('Enter your name:')
country = input('Enter your country:')
print('Name:',name)
print('Country:',country)
```

```python
#integer input
age = int(input('Enter your age:'))
print('You are',age,'years old.')
```

```python
#Float input
height = float(input('Enter your height in metres:'))
print('Your height is',height,'metres.')
```

```python
#Adding two numbers
num1 = int(input('Enter first number:'))
num2 = int(input('Enter second number:'))
sum = num1 + num2
print('The sum is',sum)
           
```

```python
age = int(input('Enter your age:'))
future_age = age + 10
print('In 10 years,you will be',future_age,'years old.')
```

```python
weight = float(input('Enter your weight (kg):'))
height = float(input('Enter your height (m):'))
bmi = weight / (height ** 2)
print('Your BMI is:',round(bmi,2))
```

```python
principal = float(input('Enter the principal amount:'))
rate = float(input('Enter the annual interest rate (%):'))
time = float(input('Enter the  time (years):'))
simple_interest = (principal * rate *time) / 100
print('The simple interest is:',simple_interest)
```

```python
math = float(input('Enter Mathematics marks:'))
english = float(input('Enter English marks:'))
science = float(input('Enter Science marks:'))
total = math + english + science
average = total / 3
print('Total marks:',total)
print('Average marks:',round(average,2))
```
