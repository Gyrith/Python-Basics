```python
#Creating custom iterators
class Count:

    def __iter__(self):
        self.num = 1
        return self

    def __next__(self):
        if self.num <= 5:
            value = self.num
            self.num += 1
            return value
        raise StopIteration

counter = Count()

for number in counter:
    print(number)
```

```python
class EvenNumbers:

    def __iter__(self):
        self.number = 2
        return self

    def __next__(self):
        if self.number <= 10:
            value = self.number
            self.number += 2
            return value
        raise StopIteration

for num in EvenNumbers():
    print(num)
```

```python
class Countdown:

    def __iter__(self):
        self.count = 3
        return self

    def __next__(self):
        if self.count > 0:
            value = self.count
            self.count -= 1
            return value
        raise StopIteration

for value in Countdown():
    print(value)
```

A generator is a special function that yield instead of return.

```python
def numbers():

    yield 1
    yield 2
    yield 3

for num in numbers():
    print(num)
```

```python
def colors():

    yield "Red"
    yield "Blue"
    yield "Green"

for color in colors():
    print(color)
```

```python
def colors():

    yield "Red"
    yield "Blue"
    yield "Green"

for color in colors():
    print(color)
```

```python
def countdown():

    yield 3
    yield 2
    yield 1

for number in countdown():
    print(number)
```

```python
#Using yield
def squares():

    for i in range(1, 6):
        yield i ** 2

for square in squares():
    print(square)
```

```python
def even_numbers():

    for i in range(2, 11, 2):
        yield i

for number in even_numbers():
    print(number)
```

```python
def greetings():

    yield "Hello"
    yield "Welcome"
    yield "Goodbye"

for word in greetings():
    print(word)
```

```python
#Generator expressions
numbers = (x ** 2 for x in range(1, 6))

for num in numbers:
    print(num)
```

```python
even = (x for x in range(20) if x % 2 == 0)

print(list(even))
```

```python
words = ["python", "java", "c++"]

upper = (word.upper() for word in words)

print(list(upper))
```

```python
#practical applications
def countdown(start):

    while start > 0:
        yield start
        start -= 1

for number in countdown(5):
    print(number)
```

```python
def countdown(start):

    while start > 0:
        yield start
        start -= 1

for number in countdown(5):
    print(number)
```

```python
numbers = (x * 10 for x in range(1, 6))

print(list(numbers))
```

```python
def fibonacci(n):

    a, b = 0, 1

    for _ in range(n):
        yield a
        a, b = b, a + b

print(list(fibonacci(10)))
```

```python
def table(number):

    for i in range(1, 11):
        yield f"{number} x {i} = {number * i}"

for row in table(7):
    print(row)
```

```python
students = ["Grace", "John", "Mary", "Alice"]

generator = (student.upper() for student in students)

print(list(generator))
```

# Lessons
 -An iterator lets python access items one at a time.
 -'iter()' creates an iterator.
 -'next()' retrieves the next item.
 -Custom iterators use '__iter__()' and '__next__()'.
 -Generators use 'yield' to produce values one by one.
 -Generator expressions are a compact way to create generators.
