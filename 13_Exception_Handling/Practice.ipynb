```python
#Understanding exceptions
number = 10
print(number / 0)
```

this raises ZeroDivisionError

```python
#invalid number
age = int('twenty')
age
```

```python
#missing list index
numbers = [10,20,30,40]
print(numbers[5])
```

This raises IndexError. 

```python
#try and except
try:
    number = 10/0
except ZeroDivisionError:
    print('You cannot divide by zero.')
```

```python
try:
    age = int(input('Enter your age:'))
    print(age)
except ValueError:
    print('Please enter a valid number.')
```

```python
try:
    numbers = [10,20,30]
    print(numbers[5])
except IndexError:
    print('Index out of range.')
```

```python
#else
try:
    number = int(input('Enter a number:'))
except ValueError:
    print('Invalid input')
else:
    print('You entered:',number)
```

```python
try:
    result = 20 / 4
except ZeroDivisionError:
    print("Cannot divide.")
else:
    print("Result:", result)
```

```python
try:
    name = input("Enter your name: ")
except Exception:
    print("Something went wrong.")
else:
    print("Welcome,", name)
```

```python
#finally
try:
    print(10/2)
finally:
    print('Program finished.')
```

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Division by zero.")
finally:
    print("This always runs.")
```

```python
try:
    age = int(input("Age: "))
    print(age)
except Exception as e:
    print("Error:", e)
```

```python
#Raising exceptions
age = -5
if age < 0:
    raise ValueError('Age cannot be negative')
```

```python
password = "123"

if len(password) < 6:
    raise Exception("Password is too short.")
```

```python
marks = 120

if marks > 100:
    raise ValueError("Marks cannot exceed 100.")
```

```python
#Custom exceptions
class InvalidAgeError(Exception):
    pass
```

```python
age = -2

try:
    if age < 0:
        raise InvalidAgeError("Invalid age entered.")
except InvalidAgeError as e:
    print(e)
```

```python
class NegativeNumberError(Exception):
    pass

number = -10

try:
    if number < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")
except NegativeNumberError as e:
    print(e)
```

```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    print("Result:", num1 / num2)

except ZeroDivisionError:
    print("Cannot divide by zero.")

except ValueError:
    print("Please enter valid numbers.")
```

```python
try:
    marks = int(input("Enter marks: "))

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    print("Marks recorded:", marks)

except ValueError as e:
    print(e)
```

```python
try:
    with open("students.txt", "r") as file:
        print(file.read())

except FileNotFoundError:
    print("The file does not exist.")
```

# Lessons
 -Exceptions prevent programs from crashing.
 -'try' tests code that may fail.
 -'except' handles errors.
 -'else' runs when there is no exception.
 -'finally' always executes.
 -I can raise my owm exceptions.
 -I can create custom exception classes.

# Common mistakes
 -Using 'except:' for every error instead of specific exceptions.
 -Forgetting to indent code inside 'try' and 'except'.
 -Raising an exception without a meaningful message.
 -Forgetting that 'finally' runs whether or not an exception occurs.
