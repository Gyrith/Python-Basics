```python
#what is a function
def greet():
    print('Hello!')
greet()
```

```python
def welcome():
    print('Welcome to Python.')
welcome()
```

```python
def goodbye():
    print('Goodbye!')
goodbye()
```

```python
#Function parameters
def greet(name):
    print('Hello,',name)
greet('Grace')
```

```python
def square(number):
    print(number ** 2)
square(5)
```

```python
def add(a,b):
    print(a + b)
add(10,20)
```

```python
#Return statement
def add(a,b):
    return a + b
result = add(5,7)
print(result)
```

```python
def multiply(a,b):
    return a*b
print(multiply(6,4))
```

```python
def area(length,width):
    return length * width
print(area(10,5))
```

```python
#Default parameters
def greet(name = 'Guest'):
    print('Hello,',name)
greet()
```

```python
def power(number,exponent=2):
    print(number ** exponent)
power(5)
```

```python
def country(name='Kenya'):
    print(name)
country()
```

```python
#Keyword Arguments
def student(name,age):
    print(name,age)
student(age=22,name='Grace')
```

```python
def rectangle(length,width):
    print(length * width)
rectangle(width=4,length=8)
```

```python
def employee(name,department):
    print(name,department)
employee(department='IT',name='James')
          
```

```python
#Variable-Length Arguments
def add_numbers(*numbers):
    print(sum(numbers))
add_numbers(2,4,6)
```

```python
def display(*names):
    for name in names:
        print(name)
display('Ann','John','Peter')
```

```python
def largest(*numbers):
    print(max(numbers))
largest(10,50,30,80)
```

```python
#Keyword Variable Arguments
def person(**details):
    print(details)
person(name='Grace',age=22)
```

```python
def student(**info):
    for key,value in info.items():
        print(key,':',value)
student(name='John',course='Data Science')
```

```python
def book(**details):
    print(details['title'])
book(title='Python Basics',author='Grace')
```

```python
def bmi(weight,height):
    return weight/(height **2)
print(round(bmi(70,1.75),2))
```

```python
def calculate_grade(mark):
    if mark >= 80:
        return 'A'
    elif mark >= 70:
        return 'B'
    elif mark >= 60:
        return 'C'
    elif mark >= 50:
        return 'D'
    else:
        return 'F'
print(calculate_grade(76))
```

```python
def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        if b != 0:
            return a / b
        return "Cannot divide by zero."
    else:
        return "Invalid operation."

print(calculator(10, 5, "+"))
print(calculator(10, 5, "*"))
```

## Lessons
 -Functions help avoid repeating code.
 -Parameters allow functions to receive input,
 -'return' sends a value back from a function.
 -Default parameters provide fallback values.
 -'*args' accepts multiple positional arguments.
 -'**kwargs' accepts multiple keyword arguments.
