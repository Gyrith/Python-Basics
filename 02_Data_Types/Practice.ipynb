```python
#int
age = 23
print(age)
print(type(age))
```

```python
#float
height = 1.75
print(height)
print(type(height))
```

```python
#str
name = 'Joy'
print(name)
print(type(name))
```

```python
#bool
Is_student = True
print(Is_student)
print(type(Is_student))
```

```python
#Type conversion
#int to float
int_num = 5
float_num = float(int_num)
print(float_num)
```

```python
#float to int
num = 3.14
int_num = int(num)
print(int_num)
```

```python
#int to str
num = 25
str_num = str(num)
print(str_num)
```

```python
#None type
var = None
print(type(var))
```

```python
#checking types
value = 10.5
if isinstance(value,float):
    print(f'{value} is a float')
```
