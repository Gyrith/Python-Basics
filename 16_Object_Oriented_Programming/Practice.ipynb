```python
#Constructors
class Student:

    def __init__ (self,name,age):
        self.name = name
        self.age = age
student1 = Student('Grace',23)
student1.name
student1.age
    
```

```python
class Calculator:

    def add(self, a, b):
        return a + b

calc = Calculator()

print(calc.add(5, 8))
```

```python
class Rectangle:

    def area(self, length, width):
        return length * width

rect = Rectangle()

print(rect.area(8, 4))
```

```python
#Inheritance
class Animal:

    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    pass

dog = Dog()

dog.speak()
```

```python
class Vehicle:

    def move(self):
        print("Vehicle is moving")

class Car(Vehicle):
    pass

car = Car()

car.move()
```

```python
class Person:

    def introduce(self):
        print("Hello!")

class Student(Person):
    pass

student = Student()

student.introduce()
```

```python
#Polymorphisim
class Dog:

    def sound(self):
        print("Bark")

class Cat:

    def sound(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

```python
class Bird:

    def move(self):
        print("Flying")

class Fish:

    def move(self):
        print("Swimming")

for animal in [Bird(), Fish()]:
    animal.move()
```

```python
class Square:

    def area(self):
        return 25

class Circle:

    def area(self):
        return 78.5

for shape in [Square(), Circle()]:
    print(shape.area())
```

```python
#Encapsulation
class Student:

    def __init__(self, name):
        self.__name = name

student = Student("Grace")

# print(student.__name)  # This will raise an error
```

```python
class Student:

    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

student = Student("Grace")

print(student.get_name())
```

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)

print(account.get_balance())
```

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"{self.name}: {self.marks}")

student1 = Student("Grace", 90)
student2 = Student("John", 85)

student1.display()
student2.display()
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_salary(self):
        return self.salary * 12

employee = Employee("Alice", 5000)

print(employee.annual_salary())
```

```python
class Book:

    def __init__(self, title, author):
        self.title = title
        self.author = author

    def details(self):
        print(f"{self.title} by {self.author}")

book = Book("Python Programming", "John Doe")

book.details()
```

# Lessons
 -A class is a blueprint for creating objects.
 -Objects are instances of a class.
 -The '__init__' method initializes object attributes.
 -Methods define what an object can do.
 -Inheritance allows one class to reuse another class's features.
 -Polymorphism lets different classes use methods with the same name.
 -Encapsulation protects object data.

# Common mistakes
 -Forgetting to use 'self' in methods.
 -Misspelling '__init__'.
 -Creating an object without the required arguments.
 -Forgetting to instantiate a class before calling its methods.
 -Trying to access private attributes directly.
