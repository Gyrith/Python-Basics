```python
#Creating dictionaries
student = {
    'name':'Grace',
    'age':23,
    'course':'Data Science'
}
student
```

```python
car = {
    'brand' : 'Toyota',
    'model' : 'Corolla',
    'year' : 2020
}
car
```

```python
book = {
    'title' : 'Python Basics',
    'author' : 'John Doe',
    'pages' : 250
}
print(book)
```

```python
#Accessing dictionary values
student = {
    'name':'Grace',
    'age':23
}
student['name']
```

```python
student['age']
```

```python
student.get('course','Not Found')
```

```python
#Modifying dictionaries
student = {
    'name' : 'Grace',
    'age' : 23
}
student['age'] = 24
student
```

```python
student['country'] = 'Kenya'
student
```

```python
student.update({'course':'Data Science'})
student
```

```python
#Dictionary methods
student = {'name': 'Grace', 'age': 24, 'country': 'Kenya', 'course': 'Data Science'}
student.keys()
```

```python
student.values()
```

```python
student.items()
```

```python
# Looping through dictionaries
for key in student:
    print(key)
```

```python
for key, value in student.items():
    print(key, ":", value)
```

```python
#Nested dictionaries
students = {
    "student1": {
        "name": "Grace",
        "age": 23
    },
    "student2": {
        "name": "John",
        "age": 21
    }
}

print(students)
```

```python
students["student1"]
```

```python
students["student2"]["name"]
```

```python
#Dictionary comprehensions
numbers = {x: x**2 for x in range(1,6)}
numbers
```

```python
even = {x: x for x in range(1, 11) if x % 2 == 0}

even
```

```python
letters = {x: len(x) for x in ["apple", "banana", "orange"]}

letters
```

```python
students = {
    "S001": {"name": "Grace", "marks": 85},
    "S002": {"name": "John", "marks": 78},
    "S003": {"name": "Mary", "marks": 92}
}

for student_id, details in students.items():
    print(student_id)
    print("Name:", details["name"])
    print("Marks:", details["marks"])
    print()
```

```python
inventory = {
    "Laptop": 5,
    "Mouse": 12,
    "Keyboard": 8
}

inventory["Monitor"] = 4

for item, quantity in inventory.items():
    print(f"{item}: {quantity}")
```

```python
sentence = input("Enter a sentence: ")

word_count = {}

for word in sentence.split():
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
```

# Lessons
 -Dictionaries store data as key-value pairs.
 -Values are accessed using keys.
 -Dictionaries are mutable.
 -I can loop through keys,values or key-vaue pairs.
 -Dictionaries are mutable.
 -Dictionary comprehensions create dictionaries efficiently.

# Common mistakes
 -Using duplicate keys (the last value replaces earlier ones).
 -Accessing a key hat doesn't exist without usin 'get()'.
 -Confusing dictionaries '{}' with lists'[]'.
 -Forgetting that dictionary keys must be unique.
