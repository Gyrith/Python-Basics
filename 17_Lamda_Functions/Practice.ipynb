# Lamda function
## Objectives
 -What is a lamda function.
 -Lamda with one argument
 -Lamda with multiple arguments.
 -Using lamda with map()
 -Using lamda with filter().
 -Using lamda with sorted().
 -Practical applications.

```python
square = lambda x: x ** 2
square(5)
```

```python
double = lambda x:x**2
double(8)
```

```python
greet = lambda name: f"Hello, {name}!"

greet("Grace")
```

```python
#Lamda with one argument
cube = lambda x: x**3
cube(4)
```

```python
length = lambda word: len(word)
length("Python")
```

```python
even = lambda x: x % 2 == 0
even(12)
```

```python
#Lamda with multiple arguments
add = lambda a,b:a+b
add(10,20)
```

```python
multiply = lambda a, b: a * b
multiply(6, 7)
```

```python
maximum = lambda a, b: a if a > b else b
maximum(18, 12)
```

```python
#Using lambda using map()
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))
squares
```

```python
numbers = [10, 20, 30]

doubles = list(map(lambda x: x * 2, numbers))
doubles
```

```python
names = ["grace", "john", "mary"]

capitalized = list(map(lambda name: name.capitalize(), names))
capitalized
```

```python
#Using lambda with filter()
numbers = [1, 2, 3, 4, 5, 6]

evens = list(filter(lambda x: x % 2 == 0, numbers))
evens
```

```python
numbers = [15, 22, 8, 41, 10]

greater = list(filter(lambda x: x > 10, numbers))
greater
```

```python
names = ["Grace", "Ann", "Christopher", "John"]

long_names = list(filter(lambda name: len(name) > 4, names))
long_names
```

```python
#Using lambda with sorted()
numbers = [5, 2, 9, 1, 7]
sorted(numbers)
```

```python
students = [("Grace", 80), ("John", 95), ("Mary", 88)]

sorted_students = sorted(students, key=lambda student: student[1])
sorted_students
```

```python
words = ["banana", "kiwi", "apple", "watermelon"]

sorted_words = sorted(words, key=lambda word: len(word))
sorted_words
```

```python
#practical applications
prices = [250, 500, 750]

discounted = list(map(lambda price: price * 0.9, prices))
discounted
```

```python
marks = [45, 78, 91, 39, 66]

passed = list(filter(lambda mark: mark >= 50, marks))
passed
```

```python
employees = [
    ("Alice", 5000),
    ("Bob", 7000),
    ("Carol", 6000)
]

highest = sorted(employees, key=lambda employee: employee[1], reverse=True)

highest
```

```python
temperatures = [20, 25, 30, 35]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, temperatures))
fahrenheit
```

```python
ages = [12, 18, 21, 15, 30, 16]

adults = list(filter(lambda age: age >= 18, ages))
adults
```

```python
products = [
    ("Laptop", 850),
    ("Mouse", 20),
    ("Keyboard", 45),
    ("Monitor", 200)
]

sorted_products = sorted(products, key=lambda product: product[1])
sorted_products
```

# Lessons
 -Lambda functions are anonymous functions.
 -They are useful for short,simple operations.
 -'map()' applies a function to every item in a collection.
 -'filter()' selects items that satisfy a condition.
 -'sorted()' can sort data using a custom key.
 -Lamda functions make code concise and readable.

## Common mistakes
 -Trying to write multiple statements inside a lambda function.
 -Forgetting to convert 'map()' or 'filter()' results to a list when needed.
 -Using lambda for complex logic that would be clearer as a regular function.
 -Forgetting the 'key=' argument when using 'sorted()' with a lambda.
