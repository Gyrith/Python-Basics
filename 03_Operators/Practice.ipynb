```python
#Arithmethic operators
a= 10
b = 3
print(a + b) #addition
print(a - b) #subtraction
print(a * b) #multiplication
print(a / b) #division
print(a // b) #floor division
print(a % b) #Modulus
print(a ** b) #Exponention
```

```python
#Comparison operators
x = 5 
y = 10
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
#Assignment operators 
num = 10
num += 5
print(num)

num -= 2
print(num)

num += 3
print(num)

num *= 3
print(num)
```

```python
#Logical operators
age = 20 
student = True
print(age >= 18 and student)
print(age >= 18 or student)
print(not student)
```

```python
#Bitwise operators
a = 10
b = 4
print('Bitwise AND: ' ,a & b)
print('Bitwise OR: ',a | b)
print('Bitwise XOR:',a^b)
print('Bitwise NOT:',~a)
print('Left Shift:', a << 1)
print('Right Shift', a >> 1)
```

```python
#Identity operators ~check whether two variables refer to the same object
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```

```python
#Identity operators
list1 = [1,2,3]
list2 = list1
list3 = [1,2,3]
print(list1 is list2)
print(list1 is list3)
print(list1 is not list3)
```

```python
#Membership operators ~check whether a value exists in a sequence
fruits = ['Apple','Banana','Orange']
print('Apple' in fruits)
print('Mango' in fruits)
print('Mango' not in fruits)
```

```python
num = [10,20,30,40,50]
print(20 in num)
print(60 in num)
print(60 not in num)
```

# Summary
-**Arithmetic operators perform mathematical operations ('+','-','*','/','//','%','**').
-**Comparison operators compare values('==','>=','<=','!=','>','<').
-**Logical operators combine boolean values('and','or','not').
-**Assignment operators assign values with operators ('+=','-=','*=',etc).
-**Bitwise operators perform bitlevel opearors(\'&','','^','-','<<','>>').
-**Identity operators check object identity('is','is not').
-**Membership operators check membership in sequences('in','not in').
