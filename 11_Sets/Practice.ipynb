```python
#Accessing set elements
fruits = {'Apple','Banana','Orange'}
fruits
```

```python
print('Apple' in fruits)
```

```python
'Mango' not in fruits
```

```python
#Adding and removing elements 
fruits = {'Apple','Banana'}
fruits.add('Orange')
fruits
```

```python
fruits.remove('Banana')
fruits
```

```python
fruits.discard('Mango')
fruits
#discard does not raise error if the item is missing.
```

```python
#Set methods
num = {1,2,3,4,5}
new_numbers = numbers.copy()
new_numbers
```

```python
numbers.clear()
numbers
```

```python
colors = {'Red','Blue','Green'}
new = colors.pop()
new
# pop() removes random element because sets are unordered.
```

```python
#Set operations##union
set1 = {1,2,3,4}
set2 = {4,7,8,9}
set1 | set2
```

```python
##intersection
print(set1 & set2)
```

```python
#difference
print(set1 - set2)
```

```python
#Looping through sets
colors = {'Red','Blue','Yellow','Green'}
for color in colors:
    print(color)
```

```python
numbers = {10, 20, 30}

for number in numbers:
    print(number * 2)
```

```python
letters = {"A", "B", "C"}

for letter in letters:
    print(letter.lower())
```

```python
#Set comprehensions
numbers ={x for x in range(1,6)}
numbers
```

```python
squares = {x**2 for x in range(1, 6)}
squares
```

```python
evens = {x for x in range(1, 21) if x % 2 == 0}
evens
```

```python
#remove duplicates
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6]
unique_numbers = set(numbers)
unique_numbers
```

```python
math = {"Grace", "John", "Mary"}
science = {"John", "Mary", "Peter"}

print("All Students:", math | science)
print("Both Subjects:", math & science)
print("Math Only:", math - science)
```

```python
word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

common = set(word1) & set(word2)

print("Common letters:", common)
```

# Lessons
 -Sets store unique values.
 -Sets are unordered.
 -I can add and remove elements from a set.
 -Set operations help compare collections.
 -Set comprehensions provide a concise way to create sets.

# Common mistakes
 -Trying to access set elements using indexes.
 -Forgetting that duplicates values are removed automatically.
 -Using 'remove()' on a value that doesn't exist.
 -Expecting sets to keep items in the order they were added.
