```python
#create file
file = open('sample.text','w')
file.write('Hello Python!')
file.close()
```

```python
file = open("student.txt", "w")
file.close()

print("student.txt created successfully.")
```

```python
file = open("notes.txt", "w")
file.close()

print("notes.txt created successfully.")
```

Opening a file with 'w' creates it if it doesn't exist. If it already exists,its content will be erased.

```python
# Writing to files
file = open("notes.txt", "w")
file.close()

print("notes.txt created successfully.")
```

```python
file = open("student.txt", "w")

file.write("Name: Grace\n")
file.write("Course: Data Science")

file.close()
```

```python
file = open("numbers.txt", "w")

for i in range(1, 6):
    file.write(str(i) + "\n")

file.close()
file
```

```python
#Reading files
file = open("sample.txt", "r")

print(file.read())

file.close()
file
```

```python
file = open("student.txt", "r")

print(file.readline())

file.close()
file
```

```python
file = open("numbers.txt", "r")

for line in file:
    print(line.strip())

file.close()
```

```python
#Appending to files
file = open("sample.txt", "a")

file.write("\nWelcome to File Handling.")

file.close()
file
```

```python
file = open("student.txt", "a")

file.write("\nCountry: Kenya")

file.close()
file
```

```python
file = open("numbers.txt", "a")

file.write("6\n")

file.close()
file
```

```python
#Using the with statement
with open("sample.txt", "r") as file:
    print(file.read())
```

```python
with open("message.txt", "w") as file:
    file.write("Learning Python File Handling.")
```

```python
with open("message.txt", "a") as file:
    file.write("\nThis file was updated.")
```

The with statement automatically closes the file.

```python
#File methods
with open("sample.txt", "r") as file:
    print(file.read())
```

```python
#readline
with open("student.txt", "r") as file:
    print(file.readline())
```

```python
with open("numbers.txt", "r") as file:
    print(file.readlines())
```

```python
#Checking if a file exists
import os

if os.path.exists("sample.txt"):
    print("sample.txt exists.")
else:
    print("sample.txt does not exist.")
```

```python
import os

filename = "message.txt"

if os.path.exists(filename):
    print("The file exists.")
else:
    print("The file does not exist.")
```

```python
import os

print(os.path.exists("student.txt"))
```

```python
with open("students.txt", "w") as file:
    file.write("Grace\n")
    file.write("John\n")
    file.write("Mary\n")

with open("students.txt", "r") as file:
    print(file.read())
```

```python
#multiplication table
number = int(input("Enter a number: "))

with open("table.txt", "w") as file:
    for i in range(1, 11):
        file.write(f"{number} x {i} = {number * i}\n")

with open("table.txt", "r") as file:
    print(file.read())
```

```python
#personal profile.
name = input("Enter your name: ")
age = input("Enter your age: ")
course = input("Enter your course: ")

with open("profile.txt", "w") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
    file.write(f"Course: {course}\n")

with open("profile.txt", "r") as file:
    print(file.read())
```
