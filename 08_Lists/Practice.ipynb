```python
#Creating lists
fruits = ['Apple','Banana','Orange']
print(fruits)
```

```python
num =[10,20,30,40,50]
print(num)
```

```python
mixed = ['Grace',23,5.8,True]
print(mixed)
```

```python
#Accessing list elements
fruits = ['Apple','Banana','Orange']
print(fruits[0])
```

```python
print(mixed[3])
```

```python
print(num [1])
```

```python
#Modifying lists
fruits = ['Apple','Banana','Orange']
fruits[1] = 'Mango'
print(fruits)
```

```python
num.insert(1,100)
print(num)
```

```python
num = [1,2,3]
num.append(4)
print(num)
```

```python
#List methods
fruits = ['Apple','Banana']
fruits.append('Orange')
print(fruits)
```

```python
fruits.remove('Banana')
print(fruits)
```

```python
fruits.sort()
print(fruits)
```

```python
#Looping through lists
fruits = ['Apple','Banana','Orange']
for fruit in fruits:
    print(fruit)
```

```python
numbers = [10,20,30]
for number in numbers:
    print(number)
```

```python
names = ['Grace','John','Mary']
for name in names:
    print('Hello',name)
```

```python
#List slicing
numbers = [10,20,30,40,50]
print(numbers[1:4])
```

```python
print(numbers[2:])
```

```python
print(numbers[3:])
```

```python
#List comprehensions
numbers = [x for x in range(1,6)]
print(numbers)
```

```python
squares = [x**2 for x in range(1,6)]
print(squares)
```

```python
evens = [x for x in range(1,21) if x% 2 == 0]
print(evens)
```

```python
marks = [78,85,92,67,88]
print('Highest:',max(marks))
print('Lowest:',min(marks))
print('Average:',sum(marks) / len(marks))
```

```python
shopping = ['Milk','Bread','Sugar']
shopping.append('Eggs')
shopping.remove('Bread')
for item in shopping:
    print(item)
```

## Lessons
 -Lists store multiple values in one variable.
 -List elements are accessed using indexes.
 -Lists can be modified by adding,changing or removing items.
 -Loops makeit easy to process every item in a list.
 -Slicing extracts parts of a list.
 -List comprehensions create lists in a concise way.

## Common mistakes
 -Using an index that does not exist(IndexError).
 -Forgetting that Python starts eindexing at 0.
 -Confusing 'append()' with 'insert()'.
 -Trying to remove an item that is not in the list.
 -Forgetting the brackets '[]' when creating a list.
